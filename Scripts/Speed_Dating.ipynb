```python
#Les attributs notés après chaque rencontre le soir du "date"
scorecard = ["attr", "sinc", "intel", "fun", "amb", "shar", "match"]
```

```python
speed_dating_corr = speed_dating_clean[scorecard].corr()
```

```python
corr = (
    speed_dating_corr['match']
    .drop('match')
    .sort_values(ascending=False)
    .reset_index()
)
corr.columns = ['variable', 'correlation']
```

```python
corr['correlation'] = corr['correlation'].round(3)

corr['target'] = 'match'

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))

ax = sns.barplot(
    data=corr,
    x='variable',
    y='correlation',
    hue='variable',     
    legend=False
)

# Ajout des étiquettes de données 
ax.bar_label(ax.containers[0], fmt='%.3f', padding=3)

plt.title(
    "Corrélation des attributs avec le succès du match",
    fontweight='bold',
    fontsize=15,
    pad=13
)

plt.xlabel("Attributs")
plt.ylabel("Coefficient de corrélation")

plt.show()
```

* Evolution à travers les différentes mesures de Temps (T1, T2, T3)

```python
# Moyennes par critère et par temps, séparées par genre
# On va faire un pivot long directement

cols_t1 = ['attr1_1', 'sinc1_1', 'fun1_1', 'intel1_1', 'amb1_1', 'shar1_1']
cols_t2 = ['attr1_2', 'sinc1_2', 'fun1_2', 'intel1_2', 'amb1_2', 'shar1_2']
cols_t3 = ['attr1_3', 'sinc1_3', 'fun1_3', 'intel1_3', 'amb1_3', 'shar1_3']

criteres = ['Attractivité', 'Sincérité', 'Fun', 'Intelligence', 'Ambition', 'Similarité']
temps = ['Time 1 (avant)', 'Time 2 (après)', 'Time 3 (suivi)']

# Création du DataFrame long avec genre 

data_list = []
for gender in speed_dating_clean['gender_label'].unique():
    subset = speed_dating_clean[speed_dating_clean['gender_label'] == gender]
    pref_t1 = subset[cols_t1].mean().values
    pref_t2 = subset[cols_t2].mean().values
    pref_t3 = subset[cols_t3].mean().values
    
    for i, critere in enumerate(criteres):
        data_list.append({
            'Critère': critere,
            'Gender': gender,
            'Time 1 (avant)': pref_t1[i],
            'Time 2 (après)': pref_t2[i],
            'Time 3 (suivi)': pref_t3[i]
        })

preferences = pd.DataFrame(data_list)

# Transformation en format long
preferences_melted = preferences.melt(
    id_vars=['Critère', 'Gender'],
    var_name='Temps',
    value_name='Importance'
)

# Ordonner les temps
preferences_melted['Temps'] = pd.Categorical(
    preferences_melted['Temps'],
    categories=temps,
    ordered=True
)

#graphique 

sns.set_theme(style="whitegrid")

sns.relplot(
    data=preferences_melted,
    x='Temps',
    y='Importance',
    hue='Critère',
    col='Gender',
    kind='line',
    marker='o',
    height=5,
    aspect=1.2,
    linewidth=2.5
).set_axis_labels(
    "Moment de mesure", "Importance moyenne"
).set_titles(
    "Évolution des critères : {col_name}",
    fontweight="bold",
    fontsize=15,
    pad=13
)

plt.legend(
    bbox_to_anchor=(1.05, 1), 
    loc='upper left', 
    borderaxespad=0.,
    title='Critères',
)

plt.tight_layout()
plt.show();
```

| Critères / Genre    | Interprétation                                                                                                                                                        |
| ------------ | --------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **`Attractivité`** |  Domination de l'attractivité et inversion des critères dominants chez les femmes à T2.      |
| **`SYNTHESE `**  | Les critères idéaux (T1) ne se retrouvent pas forcément dans la réalité de la rencontre (T2) ni sur le long terme (T3). |

### Conclusion thème 2

L'analyse révèle un décalage entre les aspirations idéales et la pratique : Lors de la rencontre, l'attractivité et l'humour s'imposent comme les véritables moteurs du match, supplantant les critères intellectuels initialement déclarés.

* L'ordre de passage lors du speed-dating a t'il une influence sur le match ? 

```python
speed_dating_clean['match_bin'] = speed_dating_clean['match'].apply(lambda x: 1 if x > 0 else 0)
taux_moyen = speed_dating_clean.groupby('order')['match_bin'].mean().reset_index()

sns.lmplot(
    data=taux_moyen,
    x='order',
    y='match_bin',
    logistic=True,
    height=4,
    aspect=2.5,
    scatter_kws={'s':50, 'alpha':0.7}
)

plt.title(
    "Taux moyen de match selon l'ordre de passage",
    fontweight='bold',
    fontsize=15 ,
    pad=13
)

plt.xlabel("Ordre de passage", fontweight='bold')
plt.ylabel("Taux moyen de match", fontweight='bold')
plt.ylim(0, 0.3)  # pour montrer clairement la proportion
plt.xlim(1, 23)
plt.xticks(range(0,23,1))

plt.grid(True)
plt.show()
```

| Variables     | Interprétation                                                                                                                                                        |
